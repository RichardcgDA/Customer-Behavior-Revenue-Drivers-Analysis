### Pearson / Spearman

```python
# Calcular correlación entre variables relevantes

from scipy.stats import pearsonr

pares_numericos = [
    ('compras_mes', 'ingreso_anual'),
    ('visitas_mes', 'gasto_publicidad_dirigida')
]

for var1, var2 in pares_numericos:
    r, p = pearsonr(df[var1], df[var2])
    print(f"{var1} vs {var2}")
    print(f"  Pearson r = {r:.3f}  (p = {p:.4f})")
    print()
```

```python
# Calcular correlación entre variables relevantes
from scipy.stats import spearmanr

pares_numericos = [
    ('compras_mes', 'ingreso_anual'),
    ('visitas_mes', 'gasto_publicidad_dirigida')
]

for var1, var2 in pares_numericos:
    r, p = spearmanr(df[var1], df[var2])
    print(f"{var1} vs {var2}")
    print(f"  Spearman r = {r:.3f}  (p = {p:.4f})")
    print()
```

### Punto-biserial

```python
# Calcular correlación entre variables relevantes
from scipy.stats import pointbiserialr

pares_biserial = [
    ('abandono', 'ingreso_anual'),
    ('abandono', 'satisfaccion'),
    ('abandono', 'visitas_mes'),
    ('miembro_premium', 'ingreso_anual'),
    ('miembro_premium', 'satisfaccion'),
]

for var_binaria, var_continua in pares_biserial:
    r, p = pointbiserialr(df[var_binaria], df[var_continua])
    print(f"{var_binaria} vs {var_continua}")
    print(f"  r_biserial = {r:.3f}  (p = {p:.4f})")
    print()
```

### V de Cramér

```python
# Función para calcular V de Cramér
from scipy.stats.contingency import association
import pandas as pd

tabla_contingencia = pd.crosstab(df['miembro_premium'], df['abandono'])

v = association(tabla_contingencia, method='cramer')
print(f"Cramér's V: {v:.4f}")
```

```python
# Aplicar V de Cramér en variables relevantes
pares_categoricos = [
    ('miembro_premium', 'abandono'),
    ('tipo_dispositivo', 'abandono'),
    ('tipo_dispositivo', 'miembro_premium'),
    ('region', 'abandono'),
    ('region', 'miembro_premium'),
    ('tipo_dispositivo', 'region'),
]

for var1, var2 in pares_categoricos:
    tabla = pd.crosstab(df[var1], df[var2])
    v = association(tabla, method='cramer')
    print(f"{var1} vs {var2}")
    print(f"  Cramér's V = {v:.4f}")
    print()
```

**Diagnóstico — Cramér's V**

- miembro_premium vs abandono: 0.1205 la asociación más fuerte del grupo, aunque sigue siendo débil.
- tipo_dispositivo vs abandono: 0.0072 prácticamente nula.
- tipo_dispositivo vs miembro_premium: 0.0197 prácticamente nula.
- region vs abandono: 0.0154 prácticamente nula.
- region vs miembro_premium: 0.0126 prácticamente nula.
- tipo_dispositivo vs region: 0.0124 prácticamente nula (confirma que no están capturando la misma información).

**Hallazgo 1 — Colinealidad fuerte entre `compras_mes` e `ingreso_anual`**

- Evidencia visual: El scatterplot muestra una relación prácticamente lineal, con los puntos formando una tendencia ascendente clara y consistente en cada nivel de compras_mes.
- Evidencia numérica: Pearson r=0.967, Spearman r=0.967 (heatmap general también lo confirma en 0.97).
- Interpretación: Ambas variables capturan esencialmente la misma información — es probable que ingreso_anual esté matemáticamente derivado de compras_mes (ej. compras × precio promedio).
- No podemos afirmar: Que exista una relación causal directa (compras_mes → ingreso_anual); solo confirmamos una asociación lineal muy fuerte, no el mecanismo que la genera.
- Implicación de negocio: Si se construye un modelo predictivo (ej. de churn o valor de cliente), incluir ambas variables generaría multicolinealidad severa y coeficientes inestables. Se recomienda usar solo una, o crear una métrica derivada (ej. ingreso promedio por compra) que aporte información no redundante

**Hallazgo 2 — Ninguna variable individual predice fuertemente el abandono**

- Evidencia visual: Las distribuciones de abandono por región y tipo de dispositivo (tablas de frecuencia) no muestran diferencias marcadas entre categorías.
- Evidencia numérica: Todas las correlaciones con abandono son bajas: punto-biserial con variables continuas (|r| < 0.03), Cramér's V con categóricas (máximo 0.12 con miembro_premium, resto < 0.02).
- Interpretación: El abandono no depende de forma simple ni lineal de ninguna variable individual del dataset (ni demográfica, ni de comportamiento, ni categórica).
- No podemos afirmar: Que estas variables sean irrelevantes para predecir el churn — relaciones no lineales, interacciones entre variables, o efectos de segmentos específicos podrían seguir siendo relevantes en un modelo más complejo (ej. árboles de decisión, regresión logística con interacciones).
- Implicación de negocio: No conviene basar estrategias de retención en un solo factor (ej. "los usuarios de tablet abandonan más"). Se requiere un enfoque multivariado o de modelado predictivo (no solo correlación bivariada) para identificar combinaciones de factores que sí expliquen el churn.

**Limitaciones**
- Correlación ≠ causalidad
- Pearson y el punto-biserial solo capturan relaciones lineales; podrían existir relaciones no lineales o interacciones entre variables que no se detectan con estos métodos.
- El tamaño de muestra grande (n=15,000) hace que incluso correlaciones muy débiles resulten estadísticamente significativas (p<0.05), aunque carezcan de relevancia práctica.
- El análisis es bivariado (variable por variable); no considera el efecto combinado de múltiples variables sobre el abandono, lo cual podría ocultar patrones reales.
- No se validó si ingreso_anual está efectivamente derivado de compras_mes (relación funcional); solo se infiere por la magnitud de la correlación.
- No se identificaron outliers de forma sistemática (solo visualmente en los scatterplots), lo que podría estar afectando algunos coeficientes.

**Próximos pasos**

Paso 1 — Probar segmentación adicional

- Segmentar por combinaciones de variables (ej. tipo_dispositivo + region) para ver si el abandono varía en subgrupos específicos.
- Analizar el abandono por rangos de edad o nivel_ingreso (binning) en vez de solo su relación lineal.

Paso 2 — Modelado multivariado

- Entrenar un modelo de clasificación (ej. regresión logística o árbol de decisión) usando todas las variables en conjunto para capturar interacciones y relaciones no lineales con abandono.

- Confirmar si existe una fórmula que relacione ambas variables (ej. dividir ingreso_anual entre compras_mes y revisar si el resultado es constante).
